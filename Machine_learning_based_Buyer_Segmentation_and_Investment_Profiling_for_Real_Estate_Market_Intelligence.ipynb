```python
# ================================================================
# MACHINE LEARNING BASED BUYER SEGMENTATION AND INVESTMENT
# PROFILING FOR REAL ESTATE MARKET INTELLIGENCE
# ================================================================
#
# DATASETS:
#   clients.csv
#   properties.csv
#
# OUTPUT:
#   - EDA
#   - Buyer-level features
#   - K-Means clustering
#   - Hierarchical clustering
#   - Elbow method
#   - Silhouette score
#   - PCA visualization
#   - Cluster profiling
#   - Geographic analysis
#   - CSV output files
#
# ================================================================


# ================================================================
# 1. INSTALL LIBRARIES
# ================================================================

!pip -q install pandas numpy matplotlib seaborn scikit-learn scipy plotly


# ================================================================
# 2. IMPORT LIBRARIES
# ================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA

from scipy.cluster.hierarchy import dendrogram, linkage

from google.colab import files

import warnings
warnings.filterwarnings("ignore")

print("=" * 70)
print("REAL ESTATE BUYER SEGMENTATION PROJECT")
print("=" * 70)


# ================================================================
# 3. UPLOAD DATASETS
# ================================================================

print("\nPlease upload clients.csv and properties.csv")

uploaded = files.upload()

print("\nUploaded files:")
for filename in uploaded.keys():
    print("  -", filename)


# ================================================================
# 4. LOAD DATA
# ================================================================

clients = pd.read_csv("clients.csv")
properties = pd.read_csv("properties.csv")

print("\n" + "=" * 70)
print("DATASET INFORMATION")
print("=" * 70)

print("\nClients shape:", clients.shape)
print("Properties shape:", properties.shape)

print("\nClients columns:")
print(clients.columns.tolist())

print("\nProperties columns:")
print(properties.columns.tolist())


# ================================================================
# 5. DISPLAY INITIAL DATA
# ================================================================

print("\nFirst 5 client records:")
display(clients.head())

print("\nFirst 5 property records:")
display(properties.head())


# ================================================================
# 6. CHECK MISSING VALUES
# ================================================================

print("\n" + "=" * 70)
print("MISSING VALUE ANALYSIS")
print("=" * 70)

print("\nClient missing values:")
print(clients.isnull().sum())

print("\nProperty missing values:")
print(properties.isnull().sum())


# ================================================================
# 7. REMOVE DUPLICATES
# ================================================================

clients_clean = clients.copy()
properties_clean = properties.copy()

clients_clean = clients_clean.drop_duplicates(
    subset="client_id"
)

properties_clean = properties_clean.drop_duplicates(
    subset="listing_id"
)

print("\nAfter duplicate removal:")
print("Clients:", clients_clean.shape)
print("Properties:", properties_clean.shape)


# ================================================================
# 8. CLEAN CLIENT DATA
# ================================================================

text_columns = [
    "client_type",
    "gender",
    "country",
    "region",
    "acquisition_purpose",
    "loan_applied",
    "referral_channel"
]

for col in text_columns:

    if col in clients_clean.columns:

        clients_clean[col] = (
            clients_clean[col]
            .astype(str)
            .str.strip()
        )

# Standardize categories

if "client_type" in clients_clean.columns:
    clients_clean["client_type"] = (
        clients_clean["client_type"].str.title()
    )

if "acquisition_purpose" in clients_clean.columns:
    clients_clean["acquisition_purpose"] = (
        clients_clean["acquisition_purpose"].str.title()
    )

if "loan_applied" in clients_clean.columns:
    clients_clean["loan_applied"] = (
        clients_clean["loan_applied"].str.title()
    )

if "country" in clients_clean.columns:
    clients_clean["country"] = (
        clients_clean["country"].str.strip()
    )

if "region" in clients_clean.columns:
    clients_clean["region"] = (
        clients_clean["region"].str.strip()
    )

if "referral_channel" in clients_clean.columns:
    clients_clean["referral_channel"] = (
        clients_clean["referral_channel"].str.title()
    )


# ================================================================
# 9. CONVERT DATE OF BIRTH TO AGE
# ================================================================

clients_clean["date_of_birth"] = pd.to_datetime(
    clients_clean["date_of_birth"],
    errors="coerce"
)

reference_date = pd.Timestamp("2026-01-01")

clients_clean["age"] = (
    (reference_date - clients_clean["date_of_birth"])
    .dt.days / 365.25
)

clients_clean["age"] = clients_clean["age"].round(1)

# Remove unrealistic ages

clients_clean.loc[
    (clients_clean["age"] < 18) |
    (clients_clean["age"] > 100),
    "age"
] = np.nan

# Fill missing age

clients_clean["age"] = clients_clean["age"].fillna(
    clients_clean["age"].median()
)


# ================================================================
# 10. CLEAN SATISFACTION SCORE
# ================================================================

clients_clean["satisfaction_score"] = pd.to_numeric(
    clients_clean["satisfaction_score"],
    errors="coerce"
)

clients_clean["satisfaction_score"] = (
    clients_clean["satisfaction_score"]
    .fillna(
        clients_clean["satisfaction_score"].median()
    )
)


# ================================================================
# 11. CLEAN PROPERTY DATA
# ================================================================

properties_clean["transaction_date"] = pd.to_datetime(
    properties_clean["transaction_date"],
    errors="coerce"
)

# Convert price

properties_clean["sale_price"] = (
    properties_clean["sale_price"]
    .astype(str)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
)

properties_clean["sale_price"] = pd.to_numeric(
    properties_clean["sale_price"],
    errors="coerce"
)

# Numeric fields

properties_clean["floor_area_sqft"] = pd.to_numeric(
    properties_clean["floor_area_sqft"],
    errors="coerce"
)

properties_clean["tower_number"] = pd.to_numeric(
    properties_clean["tower_number"],
    errors="coerce"
)

properties_clean["unit_number"] = pd.to_numeric(
    properties_clean["unit_number"],
    errors="coerce"
)

# Fill numeric missing values

properties_clean["sale_price"] = (
    properties_clean["sale_price"]
    .fillna(
        properties_clean["sale_price"].median()
    )
)

properties_clean["floor_area_sqft"] = (
    properties_clean["floor_area_sqft"]
    .fillna(
        properties_clean["floor_area_sqft"].median()
    )
)


# ================================================================
# 12. CREATE SOLD INDICATOR
# ================================================================

properties_clean["is_sold"] = (
    properties_clean["listing_status"]
    .astype(str)
    .str.lower()
    .eq("sold")
    .astype(int)
)


# ================================================================
# 13. CREATE BUYER PROPERTY FEATURES
# ================================================================

property_features = (
    properties_clean
    .dropna(subset=["client_ref"])
    .groupby("client_ref")
    .agg(
        property_count=("listing_id", "count"),

        sold_property_count=("is_sold", "sum"),

        total_property_value=("sale_price", "sum"),

        average_property_price=("sale_price", "mean"),

        total_area_sqft=("floor_area_sqft", "sum"),

        average_area_sqft=("floor_area_sqft", "mean")
    )
    .reset_index()
)

property_features = property_features.rename(
    columns={
        "client_ref": "client_id"
    }
)


# ================================================================
# 14. MERGE CLIENT + PROPERTY DATA
# ================================================================

buyer_data = clients_clean.merge(
    property_features,
    on="client_id",
    how="left"
)


# ================================================================
# 15. FILL BUYERS WITHOUT PROPERTIES
# ================================================================

property_numeric_columns = [
    "property_count",
    "sold_property_count",
    "total_property_value",
    "average_property_price",
    "total_area_sqft",
    "average_area_sqft"
]

for col in property_numeric_columns:

    buyer_data[col] = (
        buyer_data[col]
        .fillna(0)
    )


# ================================================================
# 16. CREATE ADDITIONAL FEATURES
# ================================================================

buyer_data["is_investor"] = (
    buyer_data["acquisition_purpose"]
    .eq("Investment")
    .astype(int)
)

buyer_data["uses_loan"] = (
    buyer_data["loan_applied"]
    .eq("Yes")
    .astype(int)
)

buyer_data["is_company"] = (
    buyer_data["client_type"]
    .eq("Company")
    .astype(int)
)

buyer_data["sold_ratio"] = np.where(
    buyer_data["property_count"] > 0,

    buyer_data["sold_property_count"] /
    buyer_data["property_count"],

    0
)


# ================================================================
# 17. BUYER DATASET SUMMARY
# ================================================================

print("\n" + "=" * 70)
print("BUYER-LEVEL DATASET")
print("=" * 70)

print("\nBuyer dataset shape:", buyer_data.shape)

display(buyer_data.head())


# ================================================================
# 18. EDA - ACQUISITION PURPOSE
# ================================================================

plt.figure(figsize=(8, 5))

sns.countplot(
    data=buyer_data,
    x="acquisition_purpose"
)

plt.title(
    "Buyer Distribution by Acquisition Purpose"
)

plt.xlabel("Acquisition Purpose")
plt.ylabel("Number of Buyers")

plt.tight_layout()
plt.show()


# ================================================================
# 19. EDA - CLIENT TYPE
# ================================================================

plt.figure(figsize=(8, 5))

sns.countplot(
    data=buyer_data,
    x="client_type"
)

plt.title("Client Type Distribution")

plt.xlabel("Client Type")
plt.ylabel("Number of Buyers")

plt.tight_layout()
plt.show()


# ================================================================
# 20. EDA - LOAN BEHAVIOR
# ================================================================

plt.figure(figsize=(8, 5))

sns.countplot(
    data=buyer_data,
    x="loan_applied",
    hue="acquisition_purpose"
)

plt.title(
    "Loan Usage by Acquisition Purpose"
)

plt.xlabel("Loan Applied")
plt.ylabel("Number of Buyers")

plt.tight_layout()
plt.show()


# ================================================================
# 21. EDA - COUNTRY
# ================================================================

country_counts = (
    buyer_data["country"]
    .value_counts()
    .reset_index()
)

country_counts.columns = [
    "country",
    "buyers"
]

fig = px.bar(
    country_counts,
    x="country",
    y="buyers",
    title="Buyer Distribution by Country"
)

fig.show()


# ================================================================
# 22. PROPERTY VALUE DISTRIBUTION
# ================================================================

plt.figure(figsize=(10, 5))

sns.histplot(
    buyer_data["total_property_value"],
    bins=30,
    kde=True
)

plt.title(
    "Distribution of Total Property Value per Buyer"
)

plt.xlabel("Total Property Value")
plt.ylabel("Number of Buyers")

plt.tight_layout()
plt.show()


# ================================================================
# 23. CORRELATION ANALYSIS
# ================================================================

numeric_features = [
    "age",
    "satisfaction_score",
    "property_count",
    "sold_property_count",
    "total_property_value",
    "average_property_price",
    "total_area_sqft",
    "average_area_sqft",
    "sold_ratio"
]

correlation = buyer_data[
    numeric_features
].corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title(
    "Correlation Between Buyer Features"
)

plt.tight_layout()
plt.show()


# ================================================================
# 24. PREPARE FEATURES FOR CLUSTERING
# ================================================================

numeric_cluster_features = [
    "age",
    "satisfaction_score",
    "property_count",
    "sold_property_count",
    "total_property_value",
    "average_property_price",
    "total_area_sqft",
    "average_area_sqft",
    "sold_ratio"
]

categorical_cluster_features = [
    "client_type",
    "gender",
    "country",
    "region",
    "acquisition_purpose",
    "loan_applied",
    "referral_channel"
]


# ================================================================
# 25. ENCODING + SCALING
# ================================================================

try:

    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    )

except TypeError:

    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse=False
    )


preprocessor = ColumnTransformer(

    transformers=[

        (
            "numeric",
            StandardScaler(),
            numeric_cluster_features
        ),

        (
            "categorical",
            encoder,
            categorical_cluster_features
        )
    ]
)


X = preprocessor.fit_transform(
    buyer_data[
        numeric_cluster_features +
        categorical_cluster_features
    ]
)

print("\nEncoded feature matrix shape:", X.shape)


# ================================================================
# 26. ELBOW METHOD
# ================================================================

print("\n" + "=" * 70)
print("ELBOW METHOD")
print("=" * 70)

inertia = []

K_range = range(2, 11)

for k in K_range:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(X)

    inertia.append(
        model.inertia_
    )


plt.figure(figsize=(10, 6))

plt.plot(
    list(K_range),
    inertia,
    marker="o"
)

plt.title(
    "Elbow Method for Optimal Number of Clusters"
)

plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")

plt.xticks(list(K_range))

plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()


# ================================================================
# 27. SILHOUETTE SCORE
# ================================================================

print("\n" + "=" * 70)
print("SILHOUETTE SCORE")
print("=" * 70)

silhouette_scores = []

for k in K_range:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X)

    score = silhouette_score(
        X,
        labels
    )

    silhouette_scores.append(score)

    print(
        "K =", k,
        " | Silhouette Score =",
        round(score, 4)
    )


plt.figure(figsize=(10, 6))

plt.plot(
    list(K_range),
    silhouette_scores,
    marker="o"
)

plt.title(
    "Silhouette Score by Number of Clusters"
)

plt.xlabel("Number of Clusters")
plt.ylabel("Silhouette Score")

plt.xticks(list(K_range))

plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()


# ================================================================
# 28. SELECT BEST K
# ================================================================

best_k = list(K_range)[
    np.argmax(silhouette_scores)
]

best_score = max(
    silhouette_scores
)

print("\n" + "=" * 70)
print("OPTIMAL CLUSTER NUMBER")
print("=" * 70)

print(
    "Best K:",
    best_k
)

print(
    "Best Silhouette Score:",
    round(best_score, 4)
)


# ================================================================
# 29. K-MEANS FINAL MODEL
# ================================================================

kmeans_final = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=20
)

buyer_data["cluster"] = (
    kmeans_final.fit_predict(X)
)


print("\nK-Means clustering completed.")


# ================================================================
# 30. CLUSTER DISTRIBUTION
# ================================================================

cluster_counts = (
    buyer_data["cluster"]
    .value_counts()
    .sort_index()
)

print("\nCluster distribution:")
print(cluster_counts)


plt.figure(figsize=(8, 5))

sns.countplot(
    data=buyer_data,
    x="cluster"
)

plt.title(
    "Buyer Cluster Distribution"
)

plt.xlabel("Cluster")
plt.ylabel("Number of Buyers")

plt.tight_layout()
plt.show()


# ================================================================
# 31. PCA VISUALIZATION
# ================================================================

pca = PCA(
    n_components=2,
    random_state=42
)

X_pca = pca.fit_transform(X)

pca_df = pd.DataFrame(
    X_pca,
    columns=["PCA1", "PCA2"]
)

pca_df["cluster"] = (
    buyer_data["cluster"]
    .astype(str)
)

fig = px.scatter(
    pca_df,
    x="PCA1",
    y="PCA2",
    color="cluster",
    title="Buyer Segments Using PCA",
    labels={
        "cluster": "Cluster"
    }
)

fig.show()


# ================================================================
# 32. HIERARCHICAL CLUSTERING
# ================================================================

print("\n" + "=" * 70)
print("HIERARCHICAL CLUSTERING")
print("=" * 70)

# Use a sample if dataset is large
sample_size = min(
    500,
    len(X)
)

np.random.seed(42)

sample_indices = np.random.choice(
    len(X),
    size=sample_size,
    replace=False
)

X_sample = X[
    sample_indices
]

hierarchical_model = AgglomerativeClustering(
    n_clusters=best_k,
    linkage="ward"
)

hierarchical_labels = (
    hierarchical_model.fit_predict(
        X_sample
    )
)

hierarchical_score = silhouette_score(
    X_sample,
    hierarchical_labels
)

print(
    "Hierarchical Clustering Silhouette Score:",
    round(hierarchical_score, 4)
)


# ================================================================
# 33. HIERARCHICAL DENDROGRAM
# ================================================================

print("\nGenerating hierarchical dendrogram...")

linkage_matrix = linkage(
    X_sample,
    method="ward"
)

plt.figure(figsize=(15, 7))

dendrogram(
    linkage_matrix,
    truncate_mode="lastp",
    p=30,
    leaf_rotation=90,
    leaf_font_size=10
)

plt.title(
    "Hierarchical Clustering Dendrogram"
)

plt.xlabel("Cluster / Sample")
plt.ylabel("Distance")

plt.tight_layout()
plt.show()


# ================================================================
# 34. CLUSTER PROFILE
# ================================================================

cluster_profile = (
    buyer_data
    .groupby("cluster")
    .agg(

        buyers=("client_id", "count"),

        average_age=("age", "mean"),

        average_satisfaction=(
            "satisfaction_score",
            "mean"
        ),

        investment_rate=(
            "is_investor",
            "mean"
        ),

        loan_rate=(
            "uses_loan",
            "mean"
        ),

        company_rate=(
            "is_company",
            "mean"
        ),

        average_properties=(
            "property_count",
            "mean"
        ),

        average_sold_properties=(
            "sold_property_count",
            "mean"
        ),

        average_total_value=(
            "total_property_value",
            "mean"
        ),

        average_property_price=(
            "average_property_price",
            "mean"
        ),

        average_area=(
            "average_area_sqft",
            "mean"
        ),

        average_sold_ratio=(
            "sold_ratio",
            "mean"
        )
    )
    .reset_index()
)


# Convert rates to percentages

cluster_profile["investment_rate"] *= 100
cluster_profile["loan_rate"] *= 100
cluster_profile["company_rate"] *= 100

cluster_profile = cluster_profile.round(2)


print("\n" + "=" * 70)
print("CLUSTER PROFILE")
print("=" * 70)

display(cluster_profile)


# ================================================================
# 35. PURPOSE BY CLUSTER
# ================================================================

purpose_cluster = pd.crosstab(

    buyer_data["cluster"],

    buyer_data["acquisition_purpose"],

    normalize="index"

) * 100

print("\nAcquisition Purpose by Cluster (%)")

display(
    purpose_cluster.round(2)
)


# ================================================================
# 36. LOAN BY CLUSTER
# ================================================================

loan_cluster = pd.crosstab(

    buyer_data["cluster"],

    buyer_data["loan_applied"],

    normalize="index"

) * 100

print("\nLoan Usage by Cluster (%)")

display(
    loan_cluster.round(2)
)


# ================================================================
# 37. CLIENT TYPE BY CLUSTER
# ================================================================

client_cluster = pd.crosstab(

    buyer_data["cluster"],

    buyer_data["client_type"],

    normalize="index"

) * 100

print("\nClient Type by Cluster (%)")

display(
    client_cluster.round(2)
)


# ================================================================
# 38. GENDER BY CLUSTER
# ================================================================

gender_cluster = pd.crosstab(

    buyer_data["cluster"],

    buyer_data["gender"],

    normalize="index"

) * 100

print("\nGender Distribution by Cluster (%)")

display(
    gender_cluster.round(2)
)


# ================================================================
# 39. COUNTRY BY CLUSTER
# ================================================================

country_cluster = pd.crosstab(

    buyer_data["country"],

    buyer_data["cluster"]

)

print("\nBuyer Count by Country and Cluster")

display(country_cluster)


# ================================================================
# 40. REGION BY CLUSTER
# ================================================================

region_cluster = pd.crosstab(

    buyer_data["region"],

    buyer_data["cluster"]

)

print("\nBuyer Count by Region and Cluster")

display(region_cluster.head(30))


# ================================================================
# 41. AVERAGE PROPERTY VALUE BY CLUSTER
# ================================================================

plt.figure(figsize=(10, 6))

sns.barplot(

    data=buyer_data,

    x="cluster",

    y="total_property_value",

    estimator="mean"

)

plt.title(
    "Average Property Value by Buyer Cluster"
)

plt.xlabel("Cluster")
plt.ylabel("Average Property Value")

plt.tight_layout()
plt.show()


# ================================================================
# 42. AVERAGE AGE BY CLUSTER
# ================================================================

plt.figure(figsize=(10, 6))

sns.barplot(

    data=buyer_data,

    x="cluster",

    y="age",

    estimator="mean"

)

plt.title(
    "Average Age by Buyer Cluster"
)

plt.xlabel("Cluster")
plt.ylabel("Average Age")

plt.tight_layout()
plt.show()


# ================================================================
# 43. INVESTMENT RATE BY CLUSTER
# ================================================================

investment_summary = (
    buyer_data
    .groupby("cluster")["is_investor"]
    .mean() * 100
)

plt.figure(figsize=(10, 6))

investment_summary.plot(
    kind="bar"
)

plt.title(
    "Investment Buyer Percentage by Cluster"
)

plt.xlabel("Cluster")
plt.ylabel("Investment Buyers (%)")

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ================================================================
# 44. LOAN RATE BY CLUSTER
# ================================================================

loan_summary = (
    buyer_data
    .groupby("cluster")["uses_loan"]
    .mean() * 100
)

plt.figure(figsize=(10, 6))

loan_summary.plot(
    kind="bar"
)

plt.title(
    "Loan Usage Percentage by Cluster"
)

plt.xlabel("Cluster")
plt.ylabel("Loan Users (%)")

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ================================================================
# 45. AUTOMATIC CLUSTER INTERPRETATION
# ================================================================

print("\n" + "=" * 70)
print("AUTOMATIC BUYER SEGMENT INTERPRETATION")
print("=" * 70)


for _, row in cluster_profile.iterrows():

    cluster = int(row["cluster"])

    print("\n" + "-" * 60)

    print(
        "CLUSTER",
        cluster
    )

    print(
        "Number of buyers:",
        int(row["buyers"])
    )

    print(
        "Average age:",
        round(row["average_age"], 1)
    )

    print(
        "Investment buyers:",
        round(row["investment_rate"], 1),
        "%"
    )

    print(
        "Loan users:",
        round(row["loan_rate"], 1),
        "%"
    )

    print(
        "Company buyers:",
        round(row["company_rate"], 1),
        "%"
    )

    print(
        "Average properties:",
        round(row["average_properties"], 2)
    )

    print(
        "Average property value:",
        round(
            row["average_total_value"],
            2
        )
    )

    print(
        "Average property price:",
        round(
            row["average_property_price"],
            2
        )
    )

    print(
        "Average satisfaction:",
        round(
            row["average_satisfaction"],
            2
        )
    )

    # Basic automatic interpretation

    if (
        row["company_rate"] > 50
        and
        row["average_properties"] > 1
    ):

        segment = "Corporate / Company Buyers"

    elif (
        row["investment_rate"] > 60
        and
        row["average_total_value"]
        >= cluster_profile[
            "average_total_value"
        ].median()
    ):

        segment = "Investment-Oriented Buyers"

    elif (
        row["loan_rate"] > 60
        and
        row["average_age"] < 45
    ):

        segment = "Loan-Dependent / Younger Buyers"

    elif (
        row["average_total_value"]
        >= cluster_profile[
            "average_total_value"
        ].quantile(0.75)
    ):

        segment = "High-Value Property Buyers"

    else:

        segment = "General / Mixed Buyers"

    print(
        "Suggested Segment:",
        segment
    )


# ================================================================
# 46. SAVE CLUSTER PROFILE
# ================================================================

cluster_profile.to_csv(
    "cluster_profile.csv",
    index=False
)


# ================================================================
# 47. SAVE FINAL BUYER DATA
# ================================================================

buyer_data.to_csv(
    "buyer_segmentation_results.csv",
    index=False
)


# ================================================================
# 48. SAVE COUNTRY CLUSTER ANALYSIS
# ================================================================

country_cluster.to_csv(
    "country_cluster_analysis.csv"
)


# ================================================================
# 49. SAVE REGION CLUSTER ANALYSIS
# ================================================================

region_cluster.to_csv(
    "region_cluster_analysis.csv"
)


# ================================================================
# 50. FINAL PROJECT SUMMARY
# ================================================================

print("\n")
print("=" * 70)
print("PROJECT COMPLETED SUCCESSFULLY")
print("=" * 70)

print("\nDataset:")
print("  Clients:", clients_clean.shape)
print("  Properties:", properties_clean.shape)
print("  Buyers:", buyer_data.shape)

print("\nMachine Learning:")
print("  Algorithm: K-Means")
print("  Optimal K:", best_k)
print("  Silhouette Score:", round(best_score, 4))

print("\nHierarchical Clustering:")
print(
    "  Silhouette Score:",
    round(hierarchical_score, 4)
)

print("\nOutput files created:")

print("  1. buyer_segmentation_results.csv")
print("  2. cluster_profile.csv")
print("  3. country_cluster_analysis.csv")
print("  4. region_cluster_analysis.csv")

print("\n" + "=" * 70)
print("BUYER SEGMENTATION PROJECT FINISHED")
print("=" * 70)


# ================================================================
# 51. OPTIONAL: DOWNLOAD OUTPUT FILES
# ================================================================

print("\nYou can download the output files using:")

print("""
files.download("buyer_segmentation_results.csv")
files.download("cluster_profile.csv")
files.download("country_cluster_analysis.csv")
files.download("region_cluster_analysis.csv")
""")
```
